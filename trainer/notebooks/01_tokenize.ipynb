## Tokenize the dataset

```python
#Imports
from transformers import AutoTokenizer
```

```python
tokenizer= AutoTokenizer.from_pretrained('bert-base-uncased')
example = "This is a test sentence"
BERT_input = tokenizer(example,padding='max_length', max_length = 20, 
                       truncation=True, return_tensors="pt")

print(BERT_input['input_ids'])
print(BERT_input['token_type_ids'])
print(BERT_input['attention_mask'])
```

## Explanation of the terms
- Input: [CLS]token sequence[SEP]
- Classification token is abbreviated as [CLS].
- The maximum size of tokens that can be given into the BERT model is 512. [SEP] informs BERT which token belongs to which sequence.     Therefore, if a series of tokens is fewer than 512, we can use padding to fill the empty token; otherwise, we must truncate the tokens.
- A 768-element embedding vector of each token will be the result of a BERT model. Our classifier will then receive these tokens as inputs.

```python
example_text = tokenizer.decode(BERT_input.input_ids[0])
print(example_text)
```
